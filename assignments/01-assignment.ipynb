### Question 07

Explain the difference between ASCII and Unicode. Why was Unicode developed, and what advantages does it offer over ASCII?

ASCII (American Standard Code for Information Interchange) is a character encoding system that maps numbers to characters. It uses 7 bits and is often stored in 8 bits. It can represent 128 characters, which includes English letters, digits and some basic punctuation. The key limitation for this is ASCII can only represent unaccented English characters and lacks support for most world languages.

Unicode is a universal charater encoding standard, and it is a superset of ASCII. It can represent characters from all written languages. Unicode also also represent over a trillion possible characters, far beyond ASCII’s capacity. Unicode was developed because ASCII was too limited, and the world needs one global standard that works everywhere.

### Question 08

Describe the Von Neumann architecture and its significance in modern computing. What is the Von Neumann bottleneck, and how does it affect computer performance?

The Von Neumann architecture stores program instructions and data in the same memory and uses a single pathway to transfer them to the CPU. With this invention, programs became data, which makes computers programmable and flexible. This architecture is still the foundation of almost all modern computers. Its main limitation is the Von Neumann bottleneck, where the CPU is forced to fetch instructions and data sequentially over a shared bus. This causes performance to slow down because memory access is significantly slower than CPU processing speed.
